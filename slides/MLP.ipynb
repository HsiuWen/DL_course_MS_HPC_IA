## Multilayer perceptrons, MLP

Linearity is a strong assumption and is inappropriate in many applications. The solution for nonlinearity problems include decision tree, kernel methods, etc. MLP overcomes the limitations of linear models by incorporating one or more hidden layers. We denote a minibatch of n example as $\mathbf{X} \in \mathbb{R}^{n \times d}$ where each example has d inputs (features). The first hidden layer with h hidden units is denoted as $\mathbf{H} \in \mathbb{R}^{nxh}$. In order to realize the potential of multilayer architectures, we need a key ingredient: a nonlinear activation function to be applied to each hidden unit following the affine transformation. For instance, a popular choice is the ReLU $\sigma(x) = \mathrm{max}(0, x)$ (rectified linear unit) activation function operating on its arguments elementwise.

To build more general MLPs, we can continue stacking such hidden layers, e.g.,$\mathbf{H}^{(1)} = \sigma_1(\mathbf{X} \mathbf{W}^{(1)} + \mathbf{b}^{(1)})$ and $\mathbf{H}^{(2)} = \sigma_2(\mathbf{H}^{(1)} \mathbf{W}^{(2)} + \mathbf{b}^{(2)})$ one atop another, yielding ever more expressive models


### Activation functions

Activation functions decide whether a neuron should be activated or not by calculating the weighted sum and further adding bias to it. They are differentiable operators for transforming input signals to outputs, while most of them add nonlinearity. Activation functions are fundamental to deep learning and therefore, it is important to know their properties.

- Rectified Linear unit, $\operatorname{ReLU}(x) = \max(x, 0).$ The reason for using ReLU is that its derivatives are particularly well behaved: either they vanish or they just let the argument through. This makes optimization better behaved and it mitigated the well-documented problem of vanishing gradients that plagued previous versions of neural networks. Variants: paramterized ReLU, leaking ReLU, etc.
- Losgistic Sigmoid, $\operatorname{sigmoid}(x) = \frac{1}{1 + \exp(-x)}.$. The reason for using this is that it squashes any input in the range (-inf, inf) to some value in the range (0, 1). You can think of the sigmoid as a special case of the softmax. They were the post popular activation funtion in the old time but replaced by ReLU in recent deep learing. Mush of this has to do with the fact that the optimzation is more challenges since its gradient vaniches for large positive and negative arguments. 
- Hyperbolic Tangent, $\operatorname{tanh}(x) = \frac{1 - \exp(-2x)}{1 + \exp(-2x)}.$
- GELU(x): Gaussian error linear unit
- Swish activation function: $\sigma(x) = x \operatorname{sigmoid}(\beta x)$

Predicting which activation to use in advance is usually impossible. 

### Architecture

Architecture of neural network involves how many units it should have and how these units should be connected to each other. Multilayer Feedforward neural network has chain-based architectures. The main consideration are the depth of the network and the width of each layer. A network with one hidden layer can be sufficient to fit the given training set based on the $\textbf{universal approximation theorem}$. But deeper networks are often able to use far fewer units per layer and far fewer parameters to reach the same goal. 


## Gradient-Based Learning

Designing and training a neural network is not much different from training any other machine learning model with gradient descent: specify an optimization procedure, a cost function, a model family. 
The important notice of nonlinearity of a neural network causes most interesting loss functions to become nonconvex (convex defintion: a function that line segment between any two distinct points on the graph of the function lies above or on the graph of the function between the two points). This mean the global convergence is not easy to reach since there can be many local minimuns and therefore, it is sensive to the values of initial parameters while applying stochastic gradient descent.  

Most modern nerual networks are trained using maximun likelihood where the cost function is simply the negative log-likelihood, equivalenty described as the cross-entropy between the training data and the model distribution.

### Backward propagation

The input x provides the initial information that propagates up to the hidden units at each layer and finally produces $\hat{y}$. This is called forward propagation. The back-propagation allows the information from the cost to flow backward throught the network in order to compute the gradient. 

The automatic calculation of gradients profoundly simplifies the implementation of deep learning algorithms. The field of automatic differentiation is concerned with how to compute derivatives algorithmically. Before automatic differentiation, even small changes to complicated models required recalculating complicated derivatives by hand. While we must continue to rely on automatic differentiation so we can focus on the interesting parts, you ought to know how these gradients are calculated under the hood if you want to go beyond a shallow understanding of deep learning.

### Computational graphs

To describe the back-propagation algorithm more precisely, it is helpful to use computational graph. Each node in the graph indicate a variable. An opeartion is a function of one or more variables. A graph is accompanied by a set of allowable operations. Each operation return only a single variable. If a variable y is computed by applying an operation to a variable x, then we draw a directed edge from x to y.
